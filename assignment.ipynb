```python
#plotting line chart
import matplotlib.pyplot as plt

data = {'date': pd.date_range(start='2024-01-01', periods=12, freq='M'),
        'sales': [50, 120, 300, 100, 200, 250, 320, 360, 400, 500, 600, 640]}

df = pd.DataFrame(data)
df.set_index('date', inplace=True)

df['sales'].plot.line(marker='o', color='green')
plt.title('2024 sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid('true')
plt.tight_layout
plt.show



                          
                              
```

```python
#plotting bar chart
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv('iris.csv')
data = pd.DataFrame(data)

avg_petal_length = data.groupby('variety')['petal.length'].mean()

# Create bar chart
avg_petal_length.plot.bar(color='purple')
plt.title('Average Petal Length per Species')
plt.ylabel('Petal Length (cm)')
plt.xlabel('Species')
plt.xticks(rotation=0)
plt.tight_layout()
plt.grid('true')
plt.show()
```

```python
#plotting histogram
data['sepal.length'].hist(bins=20, color='skyblue', edgecolor='black')
plt.title('Distribution of Petal Length')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Frequency')
plt.show()
```

```python
#Plotting scatter plot 
data.plot.scatter(x='sepal.length', y='petal.length')
plt.title('Sepal Length vs Petal Length')
plt.grid('true')
plt.show()
```
